# Exploring the NAQFC PM2.5 forecast datacube

The same store design as the ozone notebook, for hourly-average fine particulate matter:
`(reference_time, lead, y, x)` over CONUS, built from NOAA's GRIB2 without copying any data.

PM2.5 behaves very differently from ozone once you plot it. Ozone is a smooth regional field
spanning roughly 0–80 ppb; PM2.5 is background haze punctuated by wildfire and dust plumes, so a
single field can run from under 1 to a few thousand µg/m³. That skew drives the two places this
notebook departs from its ozone counterpart, both flagged where they occur.

```python
import icechunk
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr

# The store, matching ICECHUNK_BUCKET / ICECHUNK_PREFIX in .env_pmtf_conus
STORE_BUCKET = "naqfc"
STORE_PREFIX = "aqmv7/pm25_conus"
STORE_REGION = "us-east-1"

# Where the virtual chunks actually live
SOURCE_PREFIX = "s3://noaa-nws-naqfc-pds/"

# The variable name the GRIB parser assigns to fine particulate matter.
# NCEP local parameter (0, 13, 193); ozone is `ozcon` in the sibling store.
VAR = "pmtf"
UNITS = "PM2.5 (µg/m³)"
```

## Open the store

Two different credential scopes are in play:

- **Your IAM credentials** (`from_env=True`) read the Icechunk metadata — snapshots, manifests,
  coordinates — out of your own bucket.
- **Anonymous credentials** read the referenced GRIB messages out of NOAA's public bucket.
  Without these every chunk read fails when fetching individual chunks from the original files.

The *container* declaration — which URL prefix is virtual and how to reach it — was written into
the repository when the store was created, so it travels with the repo and a reader doesn't
redeclare it. Only the credentials, which are never persisted, have to be supplied here.

```python
repo = icechunk.Repository.open(
    storage=icechunk.s3_storage(
        bucket=STORE_BUCKET,
        prefix=STORE_PREFIX,
        region=STORE_REGION,
        from_env=True,
    ),
    authorize_virtual_chunk_access=icechunk.containers_credentials(
        {SOURCE_PREFIX: icechunk.s3_anonymous_credentials()}
    ),
)

print("branches:            ", repo.list_branches())
print("virtual containers:  ", list(repo.config.virtual_chunk_containers))
print("commits on main:     ", sum(1 for _ in repo.ancestry(branch="main")))
```

```python
cube = xr.open_zarr(
    repo.readonly_session("main").store, consolidated=False, zarr_format=3
)

# The store deliberately carries no valid_time: it is exactly
# reference_time + lead, so storing it would duplicate a 2-D coordinate that
# every append then has to keep in step. Attach it once here and it rides along
# through every isel below, so nothing downstream has to re-derive it.
# Both operands must stay DataArrays -- mixing in a raw ndarray
# (cube.lead.values) has no dimension to align on and fails to broadcast.
cube = cube.assign_coords(
    valid_time=cube.reference_time + cube.lead.astype("timedelta64[h]")
)

cube
```

## Why there is no 1-d time dimension

NAQFC runs twice a day and each run forecasts 72 hours ahead, so successive cycles re-forecast
most of the same wall-clock hours with different values. A flat `time` axis would give duplicate
labels and no way to say which forecast run a value came from.

Using **`reference_time`** (when the model was initialized) × **`lead`** (hours ahead) makes every
element unique by construction. This is also what lets the backfill write thousands of files
concurrently: each cycle owns a disjoint region, so parallel workers never contend for the same
chunk.

Valid time is not stored — it is exactly `reference_time + lead`, so keeping it would mean
maintaining a redundant 2-D coordinate on every append. It is derived above in one line.

```python
n_ref, n_lead = cube.sizes["reference_time"], cube.sizes["lead"]
logical_gb = cube[VAR].nbytes / 1e9

first, last = cube.reference_time.values[0], cube.reference_time.values[-1]
print(f"cycles span {np.datetime64(first, 'h')} .. {np.datetime64(last, 'h')}")

# The 06z and 12z runs of a day overlap in valid time
shared = np.intersect1d(cube.valid_time.values[0], cube.valid_time.values[1])
print(f"valid times shared by the first two cycles: {len(shared)} of {n_lead}")
```

## Read one forecast field

One ranged GET for a single GRIB message, decoded on read by the `gribberish` codec. Chunks span
the full grid, so any read touching one point still pulls that whole message — about 0.7 MB
compressed, 12 MB decoded. (PM2.5 messages compress a little better than ozone's ~1.2 MB, since
much of the field is near-uniform background.)

```python
REF, LEAD = -1, 0  # most recent cycle, +1 h

pmtf = cube[VAR].isel(reference_time=REF, lead=LEAD).load()
init = np.datetime64(cube.reference_time.values[REF], "h")
valid = cube.valid_time.values[REF, LEAD]

print(f"Initialized {init}, +{int(cube.lead.values[LEAD])} h -> valid {valid}")

# Percentiles, not just min/mean/max: PM2.5 is heavily right-skewed, and the
# gap between the 98th percentile and the maximum is usually a smoke or dust
# plume covering a small fraction of the grid.
v = pmtf.values
print(
    f"PM2.5 ug/m3: min={np.nanmin(v):.2f}  mean={np.nanmean(v):.2f}  "
    f"p50={np.nanpercentile(v, 50):.2f}  p98={np.nanpercentile(v, 98):.2f}  "
    f"max={np.nanmax(v):.2f}"
)
print(f"max is {np.nanmax(v) / np.nanpercentile(v, 98):.0f}x the 98th percentile")
```

```python
fig, ax = plt.subplots(figsize=(12, 7))
# robust=True clips the colour limits to the 2nd-98th percentile. Without it a
# single plume sets vmax in the thousands and the whole map washes out to one
# flat colour -- the ozone notebook needs no such thing, since its field spans
# well under one order of magnitude.
pmtf.plot(
    x="longitude",
    y="latitude",
    cmap="magma",
    robust=True,
    ax=ax,
    cbar_kwargs={"label": f"{UNITS}, 2nd-98th percentile"},
)
ax.set_title(f"NAQFC surface PM2.5 -- init {init}, valid {np.datetime64(valid, 'h')}")
plt.show()
```

## How are forecasts changing as we approach the forecasted time?

Because every run's view is retained rather than overwritten, you can ask a question a flat time
series cannot answer: **how did successive forecasts of the same hour change as that hour got
closer?**

Each point below is a different model run's prediction for one fixed valid time, plotted against
how far ahead it was forecasting. Convergence tends to be far less tidy than for ozone — a plume
that appears or shifts between runs can move a point by an order of magnitude.

```python
PY, PX = 500, 700  # 40.0 N, 102.6 W -- eastern Colorado

# Every (reference_time, lead) pair that forecast this same valid time.
target = cube.valid_time.values[-1, 0]
r, l = np.nonzero(cube.valid_time.values == target)
forecasts = cube[VAR].isel(
    reference_time=xr.DataArray(r, dims="forecast"),
    lead=xr.DataArray(l, dims="forecast"),
    y=PY, x=PX,
).sortby("lead")

fig, ax = plt.subplots(figsize=(14, 4))
# plot.line rather than plot: the generic .plot() squeezes size-1 dimensions
# before dispatching, so a valid time forecast by only one cycle -- the first
# cycle in the store, or one isolated by a gap -- collapses to a scalar and
# raises "No numeric data to plot".
forecasts.plot.line(x="lead", marker="o", ax=ax)
ax.invert_xaxis()  # time runs left to right as lead shrinks
ax.set_xlabel("lead time at which the forecast was made (h)")
ax.set_ylabel(UNITS)
ax.set_title(f"Successive forecasts of {np.datetime64(target, 'h')} at 40.0N, 102.6W")
ax.grid(alpha=0.3)
plt.show()
```

## How is the forecasted value changing with lead time?
For a single reference time, how does the forecast differ across lead times.

```python
N_LEADS = 24

series = cube[VAR].isel(reference_time=REF, lead=slice(0, N_LEADS), y=PY, x=PX)

fig, ax = plt.subplots(figsize=(11, 4))
series.plot(x="valid_time", marker="o", ax=ax)
ax.set_xlabel("valid time (UTC)")
ax.set_ylabel(UNITS)
ax.set_title(f"Surface PM2.5 at 40.0N, 102.6W -- init {init}, first {N_LEADS} h")
ax.grid(alpha=0.3)
fig.autofmt_xdate()
plt.show()
```
